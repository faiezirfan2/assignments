```python

data.groupby("item")["price"].mean().round(2)
```

```python
data.groupby("item")["price"].median().round(2)
```

```python
data.groupby("item")["price"].agg(["max","min","std"]).round(2)
```

```python
first_last = data.groupby("item")["price"].agg(first="first", last="last")
```

```python
# Computes the percentage change from the first price to the last price for each item

((first_last["last"] - first_last["first"]) / first_last["first"] * 100).round(2)
```
